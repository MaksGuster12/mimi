### Индексация

В NumPy работает привычная индексация Python, включая использование отрицательных индексов и срезов (slices).

````python
array = np.arange(6)
print(array) # весь массив
print(array[0]) # первый элемент
print(array[-1]) # Последний элемент
print(array[1:-1]) # срез всех элементов массива без первого и последнего
print(array[1:-1:2]) # срез всех элементов массива без первого и последнего с шагом 2
print(array[::-1]) # срез массива в обратном порядке
````

<b>Замечание:</b> Индексы и срезы в многомерных массивах не нужно разделять квадратными скобками, 
т.е. вместо <b>matrix[i][j]</b> нужно использовать <b>matrix[i, j]</b>

<b>Замечание:</b> Срезы в NumPy создают view ("вид"), а не копии данных, поэтому при изменении среза изменится и изначальный массив.

В качестве индексов можно использовать списки

````python
array = np.arange(6).reshape((2,3))
print(array)
array[((0, 1, 1), (0, 1, 2))] 
# Получим элементы с индексами (0,0) (1,1) (1,2)
````

### Расширение размерности

[Broadcasting](https://numpy.org/doc/stable/user/basics.broadcasting.html) позволяет производить операции над массивами разных, но всё-же согласованных размерностей.

![Imgur](https://i.imgur.com/tE3ZCWG.gif)

````python
2 * np.arange(1, 4)
````

Для того, чтобы numpy смог автоматически согласовать размерности, необходимо чтобы размерности либо совпадали либо имели единичный размер.

Прибавить строчку к матрице можно. То есть 1-мерный массив по умолчанию считается вектором-строкой:

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(4)
print(b.shape)
a + b
````

А вот строчку к столбцам - нет.

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
try:
    a + b
except Exception as e:
    print(e)
````

Для этого есть трюк с фиктивным расширением размерности. Используя структуру `matrix[np.newaxis, :]` или `matrix[None,:]` мы вектор размерности (n) преобразуем к размерности (1, n). Причём перевыделения памяти не происходит, создаётся view с заданной размерностью:

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
print(b[:,np.newaxis].shape)
a + b[:,np.newaxis]
````

Записи `matrix[np.newaxis, :]` или `matrix[None,:]` делают одно и то же, но запись с None короче и универсальнее, в других библиотеках, предоставляющих инструменты для работы с матрицами функционал такой-же.

````python
a = np.arange(12).reshape(3,4)
print(a.shape)
b = np.arange(3)
print(b.shape)
print(b[:,None].shape)
a + b[:,None]
````

Допустим мы хотим получит попарные произведения каждых значений из двух массивов разного размера. То есть хотим получить что-то такое: $c_{ij}=a_i*b_j,\forall i,j$. Можно написать цикл по $i$ и $j$, а можно воспользоваться трюком с расширением размерности:

````python
a = np.arange(1, 5)
b = np.arange(1, 4)
a[:, None] * b[None, :]
````

Библиотека numpy также поддерживает матричные произведения

Сделать матричное произведение можно следующими способами:
- `np.dot(a,b)`
- `a.dot(b)`
- `a @ b`

Все варианты делают одно и то же действие.

````python
a = np.array([[-1,2],[2,1]])
b = np.array([[3,4],[4,5]])

print(f"a={a}")
print(f"b={b}")
print(f"np.dot(a, b)=\n{np.dot(a,b)}")
print(f"a.dot(b)=\n{a.dot(b)}")
print(f"a @ b=\n{a @ b}")
````

Транспонировать матрицу можно с помощью `a.T` или `a.transpose()`. 

`.transpose()` предоставляет больше функциональности, позволяя переставлять размерности, а не только менять порядок на противоположный.

````python
a = np.arange(40).reshape(5,2,4)
print(a.shape)
print(a.T.shape)
print(a.transpose().shape)
print(a.transpose(1, 2, 0).shape)   # перестановка размерностей
````

numpy поддерживает удобные функции для агрегации: max, argmax, sum, mean, prod и другие.

Функции можно вызвать как для конкретного объекта, так и из библиотеки.

````python
print(f"a.max(): {a.max()}")
print(f"a.sum(): {a.sum()}")
print(f"np.sum(a): {np.sum(a)}")
# Это выдаст индекс, если матрицу развернуть в одномерный массив
print(f"a.argmax(): {a.argmax()}") 
# Это выдаст настоящее положение в матрице
print(f"координата в матрице a.argmax(): {np.unravel_index(a.argmax(), a.shape)}") 
````

Для всех операций можно указать параметр `axis` $-$ размерность по которой вычисляется операция. То есть после вычисления суммы указанная размерность `axis` пропадёт:

````python
print(f"a.shape: {a.shape}")
print(f"a.max(axis=1): {a.max(axis=0)}")
print(f"shape: {a.max(axis=0).shape}")
````

Можно передать несколько размерностей:

````python
print(f"a.shape: {a.shape}")
print(f"a.max(axis=(0, 2)): {a.max(axis=(0, 2))}")
print(f"shape: {a.max(axis=(0, 2)).shape}")
````

Также в numpy есть возможность векторизовать функции. То есть сделать их применимыми к целому массиву.

````python
def fun(x):
    return abs(x) * max(x, 5)
````

Просто так применить функцию к массиву не получится:

````python
try:
    fun(np.arange(10))
except Exception as e:
    print(e)
````

А если применить метод `np.vectorise`, то всё заработает:

````python
f = np.vectorize(fun)
f(np.arange(10, dtype=np.float32))
````

Но... Это работает долго. Это всё равно тот же самый цикл из Python, но просто чуть побыстрее.

````python
from time import time
start = time()
f(np.arange(10000000, dtype=np.float32))
print(f"Время работы: {time() - start} сек.")
````

А можно просто написать функцию, но с использованием элементов numpy. Тут вычисления уже идут в хорошо оптимизированном коде библиотеки.

````python
def fun_np(x):
    return np.abs(x) * np.fmax(x, 5) # поэлементный максимум с заданным значением
print("Результат тот же:")
print(fun_np(np.arange(10, dtype=np.float32)))
print("Но быстрее:")
start = time()
fun_np(np.arange(10000000, dtype=np.float32))
print(f"Время работы: {time() - start} сек.")
````

# Matplotlib

Это библиотека для рисования графиков 

Мы сейчас познакомимся с основными возможностями библиотеки.

В библиотеке есть 2 основных объекта $-$ Рисунок (Figure) и график (Subplot или Axes). 
На каждом Рисунке можно разместить несколько Графиков. Рисунок соответствует непосредственно получаемому изображению, которое можно сохранить в файл или увидеть в ноутбуке или в отдельном окне, если вы запускали скрипт на питоне.

````python
import matplotlib.pyplot as plt
````

Для создания Рисунка необходимо вызвать метод `plt.figure()`. Для создания Графика необходимо вызвать метод `.add_subplot()`.

````python
fig = plt.figure()
ax = fig.add_subplot()
ax.plot([1, 4], [2, 5]) # Рисуем отрезок из точки (1,2) в точку (4,5)
````

Использованный выше метод `.plot(x,y)` позволяет рисовать ломанные, задаваемые списком координат по осям x и по y. Вместо списка можно передавать и numpy массивы и столбцы из pandas и другие совместимые контейнеры.

````python
fig = plt.figure()
ax = fig.add_subplot()
x = np.linspace(-10, 10, 100)
y = np.sin(x) # функция x^2 - x
ax.plot(x, y)
````

Как бы это странно не звучало, но ломанная не обязана быть функцией, то есть однозначным отображением x в y:

````python
fig = plt.figure()
ax = fig.add_subplot()
x = np.linspace(-10, 10, 41)
x1 = np.cos(x)
x2 = x
ax.plot(x1, x2)
````

В mtplotlib для лаконичности добавлена возможность вызывать методы для отрисовки не только от самого объекта, но и напрямую из библиотеки. На самом деле от наших глаз скрыто хранение последнего использованного изображения или графика в памяти библиотеки.

````python
x = np.linspace(-10, 10, 100)
y = np.square(x) - x
plt.plot(x, y)

x = np.linspace(-10, 10, 100)
y = 50 * np.exp(-x ** 2)
plt.plot(x, y)
````

Можно создать новое изображение, тогда ломанные будут нарисованы отдельно.

````python
x = np.linspace(-10, 10, 100)
y = np.square(x) - x
plt.plot(x, y)

plt.figure()    # Создание нового изображения
x = np.linspace(-10, 10, 100)
y = 50 * np.exp(-x ** 2)
plt.plot(x, y)
````

Часто нам потребуется рисовать облака точек. Это можно сделать с помощью метода `scatter`:

````python
x = np.linspace(-10, 10, 100)
y = np.sin(x)
plt.plot(x, y, linewidth = 6, color = "black")
plt.scatter(x, np.random.normal(y,0.3,size=y.shape))
````

Конечно, графики ещё можно всячески настраивать:

````python
x = np.linspace(-10, 10, 100)
y = np.square(x) - x
plt.plot(x, y,
         linewidth = 5, # Толщина линии
         color = "magenta", # цвет
         linestyle = "dashed", # стиль линии
         )
plt.scatter(x, y+ np.random.normal(0,10,size=y.shape),
            marker = "*",   # вид точек
            color = "green",  # цвет
            s= 150) # размер точек

plt.scatter(x, y+ np.random.normal(0,10,size=y.shape),
            marker = "s",  # вид точек
            color = "black",  # цвет
            alpha = 0.1,    # прозрачность
            s= 150,  # размер точек
            zorder = 100) # порядок отрисовки элементов, элемент с большим значением будет на переднем плане
````

Подписи к осям и картинкам тоже настраиваемы:

````python
x = np.linspace(-10, 10, 100)
y = np.square(x) - x
plt.plot(x, y,
         linewidth = 5, # Толщина линии
         color = "magenta", # цвет
         linestyle = "dashed", # стиль линии
         )
plt.xlabel("Ось Икс")
plt.ylabel("Latex текст: $\log_2(a^b)$")    # Поддерживаются формулы из latex
plt.title("Настройка подписей")
plt.ylim(-1, 150)   # настройка границ по y
plt.xlim(-7, 15)   # настройка границ по x
````

Доступно рисование тепловых карт и линий уровня.

`np.meshgrid(x,y)` полезный метод при рисовании двумерных графиков. Из координат x и y создаёт решётку точек с заданными координатами. Работает так: $x$ понимается как вектор-строка (1, n), a $y$ понимается как вектор-столбец (m, 1) далее размерности обоих векторов расширяются так, чтобы получились две матрицы размера (m, n).


$$x=\begin{pmatrix}x_1\\\vdots\\x_n\end{pmatrix} \;\;y=\begin{pmatrix}y_1\\\vdots\\y_m\end{pmatrix}$$

$$X, Y= np.meshgrid(x,y)$$

$$X = \begin{pmatrix}x_1&\dots&x_n\\\vdots&\ddots&\vdots\\x_1&\dots&x_n\end{pmatrix}\;\;
 Y = \begin{pmatrix}y_1&\dots&y_1\\\vdots&\ddots&\vdots\\y_m&\dots&y_m\end{pmatrix}$$



````python
def fun(x, y):
    return np.exp(-np.square(x) - np.square(y))

x = np.linspace(-2, 2, 50)
y = np.linspace(-2, 2, 100)

x, y = np.meshgrid(x, y)

plt.gcf().set_size_inches(10,6)

ax = plt.subplot(1,3,1)
plt.pcolormesh(x, y, fun(x, y))
plt.title("function")

ax = plt.subplot(1,3,2)
plt.pcolormesh(x, y, x, cmap="Blues")   # Можно менять градиент цветов
plt.title("x coordinate")

ax = plt.subplot(1,3,3)
plt.pcolormesh(x, y, y, cmap="rainbow_r",  # Можно менять градиент цветов
               vmin=0,  # Минимальное значение для назначения цвета
               vmax=1)  # Максимальное значение для назначения цвета
plt.title("y coordinate")
````

Линии уровня:

````python
def fun(x, y):
    return np.exp(-np.square(x) - np.square(y))

x = np.linspace(-2, 2, 100)
y = np.linspace(-2, 2, 100)

x, y = np.meshgrid(x, y)

plt.gcf().set_size_inches(10,6)

plt.contour(x, y, fun(x, y), levels=np.linspace(0,1,100)) # линии уровня
````
